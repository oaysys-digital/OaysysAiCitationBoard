```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../dataset/citation_board_benchmarks.csv')
print('Dataset shape:', df.shape)
print(df.head(10))
```

```python
signal_cols = ['ai_citation_score','geo_score','brand_mention_score','seo_visibility_score','entity_authority_score','source_quality_score']
signal_avgs = df[signal_cols].mean()
colors = ['#6366f1','#3b82f6','#10b981','#f59e0b','#8b5cf6','#ef4444']
plt.figure(figsize=(12,5))
plt.bar(signal_avgs.index, signal_avgs.values, color=colors)
plt.title('Oaysys AI Citation Board - Average Signal Scores', fontsize=14, fontweight='bold')
plt.xlabel('Citation Signal')
plt.ylabel('Average Score (0-100)')
plt.xticks(rotation=20)
plt.axhline(y=70, color='green', linestyle='--', alpha=0.5, label='Healthy threshold')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
platform_cols = ['chatgpt_score','perplexity_score','google_ai_score','gemini_score','copilot_score']
platform_avgs = df[platform_cols].mean()
plt.figure(figsize=(8,5))
plt.bar(platform_avgs.index, platform_avgs.values, color=colors[:5])
plt.title('Oaysys AI Citation Board - AI Platform Coverage', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
